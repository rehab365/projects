```python
import numpy as np
from scipy import stats

rng = np.random.default_rng(1)
x = rng.normal(0, 1, 100_000)

def within(k):
    return round(float(np.mean(np.abs(x) < k)), 3)

print(within(1), within(2), within(3))     # 0.684 0.955 0.997

# القيمة النظرية الدقيقة
print(round(stats.norm.cdf(2) - stats.norm.cdf(-2), 3))   # 0.954
```

```python
import pandas as pd
from scipy import stats
df = pd.DataFrame({"salary":[40,42,48,50,52,55,60,150]})
col = df["salary"]
print(col.describe().round(1))
print(round(stats.skew(col),2))
print(stats.iqr(col))
# القراءة: المتوسط مرفوع بسبب الـ 150، والانحراف لليمين واضح (skew = 2.32). فالأنسب تقدم الوسيط و IQR بدل المتوسط والانحراف المعياري.
```
